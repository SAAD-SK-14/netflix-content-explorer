```python
import pandas as pd
import numpy as np

file_path = "/content/netflix_data/netflix_titles.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

display(df.head())
```

```python
# Make a copy so the original imported data remains unchanged
cleaned_df = df.copy()

# Fill missing text values
cleaned_df["country"] = cleaned_df["country"].fillna("Unknown")
cleaned_df["director"] = cleaned_df["director"].fillna("Not Listed")
cleaned_df["cast"] = cleaned_df["cast"].fillna("Not Listed")
cleaned_df["rating"] = cleaned_df["rating"].fillna("Not Rated")

# Remove rows where date_added is missing
cleaned_df = cleaned_df.dropna(subset=["date_added"]).copy()

# Convert date_added into a proper date
cleaned_df["date_added"] = pd.to_datetime(
    cleaned_df["date_added"],
    errors="coerce"
)

# Remove rows that could not be converted into dates
cleaned_df = cleaned_df.dropna(subset=["date_added"]).copy()

# Create year_added
cleaned_df["year_added"] = cleaned_df["date_added"].dt.year.astype(int)

print("Shape after cleaning:", cleaned_df.shape)
display(cleaned_df.head())
```

```python
print("Cleaned dataset shape:", cleaned_df.shape)

print("\nMissing values in important columns:")
print(
    cleaned_df[
        ["country", "director", "cast", "rating", "date_added", "year_added"]
    ].isnull().sum()
)

print("\nData types:")
print(cleaned_df[
    ["date_added", "release_year", "year_added"]
].dtypes)

print("\nYear added preview:")
display(cleaned_df[["date_added", "year_added"]].head())
```

## Data Cleaning

The dataset was cleaned by replacing missing country values with "Unknown", missing director and cast values with "Not Listed", and missing rating values with "Not Rated". Rows without a date-added value were removed. The date-added column was converted into a datetime format, and a new year-added column was created for yearly analysis.

```python
print("Movies vs TV Shows")
print(cleaned_df["type"].value_counts())

print("\nTitles added each year")
print(cleaned_df["year_added"].value_counts().sort_index())
```

Beginning the tests of the functions

```python
def show_content_type_counts():
    print("Count of Movies vs TV Shows")
    print(cleaned_df["type"].value_counts())


def show_titles_added_each_year():
    print("Number of titles added each year")
    yearly_counts = cleaned_df["year_added"].value_counts().sort_index()
    print(yearly_counts)
show_content_type_counts()
print()
show_titles_added_each_year()
```

```python
def show_top_10_countries():
    print("Top 10 Producing Countries")

    countries = cleaned_df["country"].str.split(",").explode()
    countries = countries.str.strip()

    top_countries = countries.value_counts().head(10)
    print(top_countries)


def show_top_5_genres():
    print("Top 5 Most Common Genres")

    genres = cleaned_df["listed_in"].str.split(",").explode()
    genres = genres.str.strip()

    top_genres = genres.value_counts().head(5)
    print(top_genres)
show_top_10_countries()

print()

show_top_5_genres()
```

```python
def search_titles_by_keyword():
    keyword = input("Enter a keyword to search in titles: ").strip()

    if keyword == "":
        print("Please enter a non-empty keyword.")
        return

    matches = cleaned_df[
        cleaned_df["title"].str.contains(
            keyword,
            case=False,
            na=False
        )
    ]

    if matches.empty:
        print(f'No titles found containing "{keyword}".')
        return

    print(f'\nTitles containing "{keyword}":')
    print(f"Total matches: {len(matches)}")

    columns_to_show = [
        "show_id",
        "type",
        "title",
        "release_year",
        "rating",
        "country"
    ]

    display(matches[columns_to_show].head(20))

    if len(matches) > 20:
        print("Only the first 20 results are displayed.")
```

```python
def show_rating_distribution():
    print("Rating Distribution")
    rating_counts = cleaned_df["rating"].value_counts()
    print(rating_counts)


def show_oldest_newest_release_year():
    print("Oldest and Newest Release Years")

    oldest_year = cleaned_df["release_year"].min()
    newest_year = cleaned_df["release_year"].max()

    print(f"\nOldest release year: {oldest_year}")
    print(f"Newest release year: {newest_year}")

    oldest_titles = cleaned_df[
        cleaned_df["release_year"] == oldest_year
    ][["title", "type", "release_year"]].head(10)

    newest_titles = cleaned_df[
        cleaned_df["release_year"] == newest_year
    ][["title", "type", "release_year"]].head(10)

    print("\nTitles from the oldest release year:")
    display(oldest_titles)

    print("\nTitles from the newest release year:")
    display(newest_titles)
show_rating_distribution()
show_oldest_newest_release_year()
```

```python
search_titles_by_keyword()
```

```python
def show_content_added_in_specific_year():
    year_input = input("Enter the year content was added: ").strip()

    if not year_input.isdigit():
        print("Please enter a valid year, such as 2020.")
        return

    year = int(year_input)

    matches = cleaned_df[cleaned_df["year_added"] == year]

    if matches.empty:
        print(f"No content was added in {year}.")
        return

    print(f"\nContent added in {year}:")
    print(f"Total titles: {len(matches)}")

    columns_to_show = [
        "show_id",
        "type",
        "title",
        "release_year",
        "rating",
        "country"
    ]

    display(matches[columns_to_show].head(30))

    if len(matches) > 30:
        print("Only the first 30 results are displayed.")
```

```python
show_content_added_in_specific_year()
```

```python
show_content_added_in_specific_year()
```

```python
def show_total_unique_countries():
    print("Total Unique Countries")

    countries = cleaned_df["country"].str.split(",").explode()
    countries = countries.str.strip()

    unique_countries = countries.dropna().unique()

    print(f"Total unique countries: {len(unique_countries)}")


def show_release_year_statistics():
    print("Release Year Summary Statistics")

    release_years = cleaned_df["release_year"].dropna().to_numpy()

    print(f"Mean: {np.mean(release_years):.2f}")
    print(f"Median: {np.median(release_years):.2f}")
    print(f"Minimum: {np.min(release_years)}")
    print(f"Maximum: {np.max(release_years)}")
    print(f"Standard deviation: {np.std(release_years):.2f}")
    print(f"Variance: {np.var(release_years):.2f}")

show_total_unique_countries()

print()

show_release_year_statistics()
```

## Interactive Menu

The menu-driven interface uses a `while True` loop to repeatedly display choices to the user. The program calls the appropriate function according to the selected option and exits safely when the user enters `0`.

Now, building a menu...

```python
def show_menu():
    print("\n" + "=" * 50)
    print("        NETFLIX CONTENT EXPLORER")
    print("=" * 50)
    print("1. Count of Movies vs TV Shows")
    print("2. Number of titles added each year")
    print("3. Top 10 producing countries")
    print("4. Top 5 most common genres")
    print("5. Rating distribution")
    print("6. Oldest and newest release years")
    print("7. Search titles by keyword")
    print("8. Content added in a specific year")
    print("9. Total unique countries")
    print("10. Summary statistics on release year")
    print("0. Quit")
    print("=" * 50)


while True:
    show_menu()

    choice = input("Enter your choice (0-10): ").strip()

    if choice == "1":
        show_content_type_counts()

    elif choice == "2":
        show_titles_added_each_year()

    elif choice == "3":
        show_top_10_countries()

    elif choice == "4":
        show_top_5_genres()

    elif choice == "5":
        show_rating_distribution()

    elif choice == "6":
        show_oldest_newest_release_year()

    elif choice == "7":
        search_titles_by_keyword()

    elif choice == "8":
        show_content_added_in_specific_year()

    elif choice == "9":
        show_total_unique_countries()

    elif choice == "10":
        show_release_year_statistics()

    elif choice == "0":
        print("Thank you for using Netflix Content Explorer.")
        break

    else:
        print("Invalid choice. Please enter a number from 0 to 10.")
```

```python
# Collect exact values for the final project insights

movie_tv_counts = cleaned_df["type"].value_counts()

countries = cleaned_df["country"].str.split(",").explode().str.strip()
top_country = countries.value_counts().index[0]
top_country_count = countries.value_counts().iloc[0]

genres = cleaned_df["listed_in"].str.split(",").explode().str.strip()
top_genre = genres.value_counts().index[0]
top_genre_count = genres.value_counts().iloc[0]

most_added_year = cleaned_df["year_added"].value_counts().idxmax()
most_added_count = cleaned_df["year_added"].value_counts().max()

oldest_year = cleaned_df["release_year"].min()
newest_year = cleaned_df["release_year"].max()

print("Movies and TV Shows:")
print(movie_tv_counts)

print(f"\nTop producing country: {top_country} ({top_country_count} titles)")
print(f"Top genre: {top_genre} ({top_genre_count} titles)")
print(f"Year with most titles added: {most_added_year} ({most_added_count} titles)")
print(f"Oldest release year: {oldest_year}")
print(f"Newest release year: {newest_year}")
```
